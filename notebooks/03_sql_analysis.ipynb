Phase 4: the analysis queries in `sql/analysis_queries.sql`, run against the database built in notebook 02. The same file feeds the dropdown on the dashboard's SQL Query page, so the two never drift apart.

```python
import sys
from pathlib import Path
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))
from nhl_pipeline import config
from nhl_pipeline.db import connect, load_named_queries

conn = connect(config.DB_PATH, read_only=True)
QUERIES = {q["id"]: q for q in load_named_queries()}

def run(qid):
    q = QUERIES[qid]
    print(f"{qid} · {q['title']}\n{q['description']}\n")
    print(q["sql"], "\n")
    return pd.read_sql_query(q["sql"], conn)

print(len(QUERIES), "queries loaded")
```
